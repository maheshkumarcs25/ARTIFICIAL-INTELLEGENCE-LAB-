```python

# 8-Puzzle Problem using Depth First Search (DFS)

GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row, col = divmod(zero, 3)

    # Up
    if row > 0:
        new_state = list(state)
        new_state[zero], new_state[zero - 3] = \
            new_state[zero - 3], new_state[zero]
        neighbors.append(("Up", tuple(new_state)))

    # Down
    if row < 2:
        new_state = list(state)
        new_state[zero], new_state[zero + 3] = \
            new_state[zero + 3], new_state[zero]
        neighbors.append(("Down", tuple(new_state)))

    # Left
    if col > 0:
        new_state = list(state)
        new_state[zero], new_state[zero - 1] = \
            new_state[zero - 1], new_state[zero]
        neighbors.append(("Left", tuple(new_state)))

    # Right
    if col < 2:
        new_state = list(state)
        new_state[zero], new_state[zero + 1] = \
            new_state[zero + 1], new_state[zero]
        neighbors.append(("Right", tuple(new_state)))

    return neighbors


def dfs(start):
    stack = [(start, [])]
    visited = set()

    while stack:
        state, path = stack.pop()

        if state == GOAL:
            return path

        if state in visited:
            continue

        visited.add(state)

        for move, next_state in get_neighbors(state):
            if next_state not in visited:
                stack.append(
                    (next_state, path + [(move, next_state)])
                )

    return None


def print_board(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])


# Initial state
start = (1, 2, 3,
         4, 5, 6,
         0, 7, 8)

print("Initial State:")
print_board(start)

solution = dfs(start)

if solution:
    print("\nSolution found!")
    print("Number of moves:", len(solution))

    for move, state in solution:
        print("\nMove:", move)
        print_board(state)
else:
    print("No solution found.")
```

```python
# 8-Puzzle Problem using Iterative Deepening Search (IDS)

GOAL = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


def get_neighbors(state):
    neighbors = []

    zero = state.index(0)
    row, col = divmod(zero, 3)

    # Up
    if row > 0:
        new_state = list(state)
        new_state[zero], new_state[zero - 3] = \
            new_state[zero - 3], new_state[zero]
        neighbors.append(("Up", tuple(new_state)))

    # Down
    if row < 2:
        new_state = list(state)
        new_state[zero], new_state[zero + 3] = \
            new_state[zero + 3], new_state[zero]
        neighbors.append(("Down", tuple(new_state)))

    # Left
    if col > 0:
        new_state = list(state)
        new_state[zero], new_state[zero - 1] = \
            new_state[zero - 1], new_state[zero]
        neighbors.append(("Left", tuple(new_state)))

    # Right
    if col < 2:
        new_state = list(state)
        new_state[zero], new_state[zero + 1] = \
            new_state[zero + 1], new_state[zero]
        neighbors.append(("Right", tuple(new_state)))

    return neighbors


def depth_limited_search(state, depth, path, visited):

    if state == GOAL:
        return path

    if depth == 0:
        return None

    visited.add(state)

    for move, next_state in get_neighbors(state):

        if next_state not in visited:

            result = depth_limited_search(
                next_state,
                depth - 1,
                path + [(move, next_state)],
                visited
            )

            if result is not None:
                return result

    visited.remove(state)

    return None


def ids(start, max_depth=30):

    for depth in range(max_depth + 1):

        print("Searching at depth:", depth)

        result = depth_limited_search(
            start,
            depth,
            [],
            set()
        )

        if result is not None:
            return result

    return None


def print_board(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])


# Initial state
start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

print("Initial State:")
print_board(start)

solution = ids(start)

if solution:
    print("\nSolution found!")
    print("Number of moves:", len(solution))

    for move, state in solution:
        print("\nMove:", move)
        print_board(state)
else:
    print("No solution found.")
```
